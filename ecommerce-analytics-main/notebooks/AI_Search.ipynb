```python
import pandas as pd

products = pd.read_csv("../ecommerce_dataset/products.csv")
```

```python
def search_products(
    category=None,
    min_price=None,
    max_price=None,
    min_rating=None
):
    result = products.copy()

    if category:
        result = result[
            result["category"].str.contains(category, case=False, na=False)
        ]

    if min_price is not None:
        result = result[result["price"] >= min_price]

    if max_price is not None:
        result = result[result["price"] <= max_price]

    if min_rating is not None:
        result = result[result["rating"] >= min_rating]

    if "rating" in result.columns:
        result = result.sort_values(
        by="rating",
        ascending=False
    )

    return result
```

```python
search_products(
    category="Electronics",
    max_price=200,
    min_rating=4
)
```

```python
def best_products(category=None, top_n=5):

    result = products.copy()

    if category:
        result = result[
            result["category"].str.contains(category, case=False, na=False)
        ]

    result = result.sort_values(
        by="rating",
        ascending=False
    )

    return result.head(top_n)
```

```python
best_products("Electronics")
```

```python
def compare_products(product1, product2):

    p1 = products[
        products["product_name"].str.lower() == product1.lower()
    ]

    p2 = products[
        products["product_name"].str.lower() == product2.lower()
    ]

    if p1.empty:
        return f"'{product1}' not found."

    if p2.empty:
        return f"'{product2}' not found."

    comparison = pd.DataFrame({
        "Feature": ["Price", "Rating", "Category", "Brand"],
        product1: [
            p1.iloc[0]["price"],
            p1.iloc[0]["rating"],
            p1.iloc[0]["category"],
            p1.iloc[0]["brand"],
        ],
        product2: [
            p2.iloc[0]["price"],
            p2.iloc[0]["rating"],
            p2.iloc[0]["category"],
            p2.iloc[0]["brand"],
        ]
    })

    return comparison
```

```python
compare_products(
    "NeoTech Help",
    "Astra Upon"
)
```

```python
def recommend_products(
    category=None,
    max_price=None,
    min_rating=None,
    top_n=5
):
    result = products.copy()

    # Filter by category
    if category:
        result = result[
            result["category"].str.contains(category, case=False, na=False)
        ]

    # Filter by price
    if max_price is not None:
        result = result[result["price"] <= max_price]

    # Filter by rating
    if min_rating is not None:
        result = result[result["rating"] >= min_rating]

    result["score"] = (
    result["rating"] * 0.8 +
    (1 / result["price"]) * 100 * 0.2
    )

    result = result.sort_values(
    by="score",
    ascending=False
    )

    return result[
    [
        "product_name",
        "category",
        "brand",
        "price",
        "rating"
    ]
].head(top_n)
```

```python
recommend_products(
    category="Electronics",
    max_price=300,
    min_rating=4
)
```

```python
def get_product(product_name):

    result = products[
        products["product_name"].str.contains(
            product_name,
            case=False,
            na=False
        )
    ]

    return result
```

```python
get_product("NeoTech")
```

```python
def compare_summary(product1, product2):

    p1 = get_product(product1).iloc[0]
    p2 = get_product(product2).iloc[0]

    summary = []

    if p1["rating"] > p2["rating"]:
        summary.append(
            f"{p1['product_name']} has a higher rating."
        )
    else:
        summary.append(
            f"{p2['product_name']} has a higher rating."
        )

    if p1["price"] < p2["price"]:
        summary.append(
            f"{p1['product_name']} is cheaper."
        )
    else:
        summary.append(
            f"{p2['product_name']} is cheaper."
        )

    if p1["brand"] == p2["brand"]:
        summary.append("Both products are from the same brand.")

    if p1["category"] == p2["category"]:
        summary.append("Both belong to the same category.")

    return summary
```

```python
compare_summary(
    "NeoTech",
    "Harbor"
)
```
